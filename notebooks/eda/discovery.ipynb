```python
import pandas as pd
import sqlalchemy
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
engine = sqlalchemy.create_engine('sqlite:///../../data/processed/flights.db')

flights = pd.read_sql_table(table_name='flights_clean', con=engine)
```

```python
# Example: assuming your data is already loaded into a DataFrame called df
# df = pd.read_csv('your_flight_data.csv')

# Set the overall style
sns.set(style="whitegrid")

# Create subplots
fig, axes = plt.subplots(3, 1, figsize=(12, 16), tight_layout=True)

# Plot 1: flight_status distribution
sns.countplot(data=flights, x='flight_status', order=flights['flight_status'].value_counts().index, ax=axes[0], palette='Blues_d')
axes[0].set_title('Distribution of Flight Status')
axes[0].set_xlabel('Flight Status')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=45)

# Plot 2: arrival_airport distribution
sns.countplot(data=flights, x='arrival_airport', order=flights['arrival_airport'].value_counts().index[:10], ax=axes[1], palette='Greens_d')
axes[1].set_title('Top 10 Arrival Airports')
axes[1].set_xlabel('Arrival Airport')
axes[1].set_ylabel('Count')
axes[1].tick_params(axis='x', rotation=45)

# Plot 3: airline_name distribution
sns.countplot(data=flights, x='airline_name', order=flights['airline_name'].value_counts().index[:10], ax=axes[2], palette='Oranges_d')
axes[2].set_title('Top 10 Airlines')
axes[2].set_xlabel('Airline Name')
axes[2].set_ylabel('Count')
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
